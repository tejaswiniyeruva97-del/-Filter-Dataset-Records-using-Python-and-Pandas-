```python
import pandas as pd
```

```python
data = {
    'Name': ['Arun', 'Priya', 'Rahul', 'Sneha', 'Kiran',
             'Divya', 'Ravi', 'Anu'],
    'Age': [20, 21, 19, 22, 20, 21, 23, 19],
    'Marks': [85, 92, 67, 78, 95, 55, 72, 88],
    'Department': ['CSE', 'ECE', 'CSE', 'EEE',
                   'CSE', 'ECE', 'EEE', 'CSE']
}

df = pd.DataFrame(data)

print("Original Dataset:")
display(df)
```

```python
high_marks = df[df['Marks'] > 80]

print("Students with Marks Above 80:")
display(high_marks)
```

```python
cse_students = df[df['Department'] == 'CSE']

print("CSE Department Students:")
display(cse_students)
```

```python
cse_students = df[df['Department'] == 'CSE']

print("CSE Department Students:")
display(cse_students)
```

```python
age_filter = df[df['Age'] > 20]

print("Students Older Than 20:")
display(age_filter)
```

```python
result_and = df[
    (df['Marks'] > 80) & (df['Age'] < 22)
]

print("Students with Marks Above 80 AND Age Below 22:")
display(result_and)
```

```python
result_or = df[
    (df['Department'] == 'CSE') |
    (df['Department'] == 'ECE')
]

print("Students from CSE OR ECE Departments:")
display(result_or)
```

```python
print("Total Records:", len(df))
print("High Marks Students:", len(high_marks))
print("CSE Students:", len(cse_students))
print("Students Older Than 20:", len(age_filter))
print("AND Condition Records:", len(result_and))
print("OR Condition Records:", len(result_or))
```

```python
print("Original Dataset")
display(df)

print("High Marks Dataset")
display(high_marks)

print("CSE Students Dataset")
display(cse_students)

print("Age Filtered Dataset")
display(age_filter)

print("AND Condition Dataset")
display(result_and)

print("OR Condition Dataset")
display(result_or)
```
